```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, roc_auc_score, roc_curve,classification_report
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import label_binarize
```

# Data Preparation

```python
# Load data
df = pd.read_csv("output_csv_files/ward_temporal_analysis.csv", parse_dates=['Month'])
# Convert 'Month' column to datetime **before** dropping NaNs
df['Month'] = pd.to_datetime(df['Month'], errors='coerce')

# Now drop rows where datetime conversion failed or other NaNs exist
df = df.dropna()

# Drop unnecessary columns
df = df.drop(columns=['Ward Name', 'Monthly_Burglaries','Ward_Area'], errors='ignore')

# Extract year and numeric month
df['Year'] = df['Month'].dt.year
df['Month_Num'] = df['Month'].dt.month  # numeric month for modeling

# Check if the years make sense
print("Years in dataset:", df['Year'].unique())

# Encode categorical Ward ID
le = LabelEncoder()
df['Ward ID'] = le.fit_transform(df['Ward ID'])

# Bin target variable into 3 categories (low, mid, high crime density)
df['Crime_Density_Level'] = pd.qcut(df['Monthly_Crime_Density'], q=3, labels=[0, 1, 2])

# Define features and target
features = ['Ward ID', 'Year', 'Month_Num']
target = 'Crime_Density_Level'

# Split into training and test set
df_train = df[df['Year'] <= 2024]
df_test = df[df['Year'] > 2024]

# Final feature matrices
X_train = df_train[features]
y_train = df_train[target].astype(int)

X_test = df_test[features]
y_test = df_test[target].astype(int)
```

# Logistic Regression Modelling and Evaluation

```python
# Initialize the logistic regression model.
# You can adjust hyperparameters such as the regularization strength if needed.
# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test data using the same scaler
X_test_scaled = scaler.transform(X_test)

log_reg = LogisticRegression(max_iter=50000,solver='lbfgs')

# Train the model on the training set.
log_reg.fit(X_train_scaled, y_train)


y_pred = log_reg.predict(X_test_scaled)
y_pred_prob = log_reg.predict_proba(X_test_scaled)  # Keep full matrix


# Binarize the output labels for multiclass AUC
y_test_bin = label_binarize(y_test, classes=[0, 1, 2])

# Compute AUC using OvR strategy
auc = roc_auc_score(y_test_bin, y_pred_prob, multi_class='ovr')
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("ROC AUC Score:", auc)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
# Suppose you have y_test (true labels) and y_pred_prob (predicted probabilities)
classes = np.unique(y_test)
n_classes = len(classes)

# One-hot encode y_test if not already
y_test_bin = label_binarize(y_test, classes=classes)

# Initialize plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.grid(True, which='major', linestyle='-', linewidth=0.5, color='#d4d4d4', zorder=0)
ax.grid(True, which='minor', linestyle='--', linewidth=0.5, color='#efefef', zorder=0)

# Colors for each class
colors = plt.cm.get_cmap('tab10', n_classes)

# Plot ROC for each class
for i in range(n_classes):
    fpr, tpr, thresholds = roc_curve(y_test_bin[:, i], y_pred_prob[:, i])
    auc = roc_auc_score(y_test_bin[:, i], y_pred_prob[:, i])
    
    ax.plot(fpr, tpr,
            color=colors(i),
            linestyle='-',
            linewidth=2,
            label=f"Class {classes[i]} (AUC={auc:.2f})",
            zorder=3)

# Diagonal reference line (random guess)
ax.plot([0, 1], [0, 1], color='#94A3B8', linestyle='--', linewidth=1.5, label="Random guess", zorder=2)

# Plot settings
ax.set_xlabel("False Positive Rate", fontsize=14, labelpad=10, fontweight='bold')
ax.set_ylabel("True Positive Rate", fontsize=14, labelpad=10, fontweight='bold')
ax.set_title("Multiclass ROC Curve (One-vs-Rest)", fontsize=14, pad=15, fontweight='bold')
ax.set_xticks(np.arange(0, 1.1, 0.2))
ax.set_yticks(np.arange(0, 1.1, 0.2))
ax.set_xticks(np.arange(0, 1.05, 0.05), minor=True)
ax.set_yticks(np.arange(0, 1.05, 0.05), minor=True)
ax.tick_params(axis='both', which='major', labelsize=10, direction='out', length=4, width=1)
ax.tick_params(axis='both', which='minor', direction='out', length=2, width=1, labelsize=0)
ax.legend(loc="lower right", framealpha=0.9, edgecolor='#CCCCCC')
ax.set_facecolor('white')

# Remove top/right spines
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
ax.spines['left'].set_linewidth(1.0)
ax.spines['bottom'].set_linewidth(1.0)
ax.spines['left'].set_color('#333333')
ax.spines['bottom'].set_color('#333333')

fig.tight_layout()
plt.show()
```

# Cross-Validation with 5 Folds

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(log_reg, X_train_scaled, y_train, cv=5)
print(f"F1-Weighted Cross-Validation Scores: {scores}")
```

# Hyperparementer Tuning on max_iter

```python
# Define the range for max_iter values
max_iter_values = np.arange(1000, 50001, 1000)
accuracy_list = []

# Loop through different max_iter values and record the accuracy
for max_iter in max_iter_values:
    log_reg = LogisticRegression(max_iter=max_iter)
    log_reg.fit(X_train, y_train)
    y_pred = log_reg.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracy_list.append(accuracy)
    
plt.figure(figsize=(10, 6))
plt.plot(max_iter_values, accuracy_list, marker='o', linestyle='-')
plt.xlabel('max_iter')
plt.ylabel('Accuracy')
plt.title('Effect of max_iter on Logistic Regression Accuracy')
plt.grid(True)
plt.show()
```
